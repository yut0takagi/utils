# パスワードがわかっている状態での複数ファイルの一括解除

```python
import glob
from pypdf import PdfReader, PdfWriter

SAVE_DIR = "src"  # 保存ディレクトリを指定
files = glob.glob(f"{SAVE_DIR}/*.pdf")

for file in files:
    print(f"Processing file: {file}")
    reader = PdfReader(file)
    
    if reader.is_encrypted:
        print("File is encrypted. Attempting to decrypt...")
        password = "enshuu"  # 必要に応じてパスワードを指定
        if not reader.decrypt(password):  # 復号化を試みる
            print(f"Failed to decrypt {file}. Skipping.")
            continue
        print("Decryption successful.")
    
    # クローン作成
    writer = PdfWriter()
    try:
        writer.clone_reader_document_root(reader)
        output_file = file.replace(".pdf", ".pdf")  # クローンファイル名を設定
        with open(output_file, "wb") as f:
            writer.write(f)
        print(f"File saved to: {output_file}")
    except Exception as e:
        print(f"Error while processing {file}: {e}")
```
